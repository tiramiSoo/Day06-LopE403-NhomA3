# Collaborative Filtering — Vietnamese Food Recommendation

Hai kỹ thuật:
- **User-based CF**: tìm user có taste tương tự → mượn rating của họ để dự đoán
- **Item-based CF**: tìm món tương tự với món user đã thích → gợi ý món chưa ăn

```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

foods = pd.read_csv(r'C:\Users\Admin\Downloads\A-New-Dataset-and-Empirical-Evaluation-for-Vietnamese-Food-Recommendation-System-main\A-New-Dataset-and-Empirical-Evaluation-for-Vietnamese-Food-Recommendation-System-main\Data\Clean Dataset\foods.csv')
ratings = pd.read_csv(r'C:\Users\Admin\Downloads\A-New-Dataset-and-Empirical-Evaluation-for-Vietnamese-Food-Recommendation-System-main\A-New-Dataset-and-Empirical-Evaluation-for-Vietnamese-Food-Recommendation-System-main\Data\Clean Dataset\ratings.csv')

# Build user-item matrix: rows=users, cols=foods, values=ratings (NaN = chưa ăn)
user_item = ratings.pivot_table(index='userId', columns='foodId', values='rating')

sparsity = user_item.isna().sum().sum() / user_item.size
print(f"Matrix shape : {user_item.shape}  ({user_item.shape[0]} users × {user_item.shape[1]} foods)")
print(f"Sparsity     : {sparsity:.1%}  (phần chưa được rate)")
print(f"Rating range : {ratings['rating'].min()} – {ratings['rating'].max()}")
user_item.head()
```

```python
## User-based Collaborative Filtering

def user_based_cf(user_item_matrix, target_user, top_k_users=10, top_n=5):
    """
    Gợi ý món ăn cho target_user dựa trên rating của những user tương tự.

    Ý tưởng:
      1. Tính cosine similarity giữa tất cả cặp user.
      2. Lấy top-K user gần nhất (neighbor).
      3. Dự đoán rating cho món target_user chưa ăn bằng weighted average
         rating của neighbors, trọng số = similarity score.
      4. Trả về top-N món có predicted rating cao nhất.
    """
    # Fill NaN = 0 để tính cosine (0 nghĩa là chưa rate, không phải ghét)
    matrix_filled = user_item_matrix.fillna(0)

    # Ma trận similarity: shape (n_users, n_users)
    sim_matrix = cosine_similarity(matrix_filled)
    sim_df = pd.DataFrame(sim_matrix, index=user_item_matrix.index, columns=user_item_matrix.index)

    # Top-K user tương tự nhất (loại bỏ bản thân)
    neighbors = sim_df[target_user].drop(target_user).nlargest(top_k_users)

    # Món target_user chưa rate
    unrated_items = user_item_matrix.loc[target_user][user_item_matrix.loc[target_user].isna()].index

    # Dự đoán rating từng món chưa ăn
    predictions = {}
    for item in unrated_items:
        neighbor_ratings = user_item_matrix.loc[neighbors.index, item].dropna()
        if neighbor_ratings.empty:
            continue
        sims = neighbors[neighbor_ratings.index]
        predictions[item] = (sims * neighbor_ratings).sum() / sims.abs().sum()

    # Map food_id -> dish_name
    pred_series = pd.Series(predictions).nlargest(top_n)
    result = foods.set_index('food_id').loc[pred_series.index, 'dish_name'].to_frame()
    result['predicted_rating'] = pred_series.values
    return result


# Demo: gợi ý cho user 0
print("=== User-based CF — Gợi ý cho user 0 ===\n")
print(user_based_cf(user_item, target_user=0, top_k_users=10, top_n=5).to_string())
```

```python
## Item-based Collaborative Filtering

def item_based_cf(user_item_matrix, target_user, top_n=5):
    """
    Gợi ý món ăn cho target_user dựa trên độ tương tự giữa các món.

    Ý tưởng:
      1. Tính cosine similarity giữa tất cả cặp món (transpose matrix).
      2. Lấy những món target_user đã rate cao (>= 3.5).
      3. Với mỗi món chưa ăn, tính score = tổng(sim × rating_đã_cho)
         weighted bởi similarity với các món đã thích.
      4. Trả về top-N món có score cao nhất.
    """
    # Cosine similarity trên item axis (transpose: shape n_items × n_users)
    matrix_filled = user_item_matrix.fillna(0)
    item_sim = cosine_similarity(matrix_filled.T)
    item_sim_df = pd.DataFrame(item_sim, index=user_item_matrix.columns, columns=user_item_matrix.columns)

    # Món user đã rate cao (>= 3.5)
    user_ratings = user_item_matrix.loc[target_user].dropna()
    liked_items = user_ratings[user_ratings >= 3.5]

    # Món chưa rate
    unrated_items = user_item_matrix.loc[target_user][user_item_matrix.loc[target_user].isna()].index

    # Score cho từng món chưa ăn
    scores = {}
    for item in unrated_items:
        sims = item_sim_df.loc[item, liked_items.index]
        denom = sims.abs().sum()
        if denom == 0:
            continue
        scores[item] = (sims * liked_items).sum() / denom

    pred_series = pd.Series(scores).nlargest(top_n)
    result = foods.set_index('food_id').loc[pred_series.index, 'dish_name'].to_frame()
    result['score'] = pred_series.values
    return result


# Demo: gợi ý cho user 0
print("=== Item-based CF — Gợi ý cho user 0 ===\n")
print(item_based_cf(user_item, target_user=0, top_n=5).to_string())
```

```python
## Evaluation Setup — Train/Test Split + Precompute Similarities

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error

# 80/20 split ở mức từng rating (không phải user)
train_ratings, test_ratings = train_test_split(ratings, test_size=0.2, random_state=42)

train_matrix = train_ratings.pivot_table(index='userId', columns='foodId', values='rating')
global_mean = train_ratings['rating'].mean()

# Precompute cả hai similarity matrix một lần — dùng lại cho mọi prediction
train_filled = train_matrix.fillna(0)
user_sim_df = pd.DataFrame(
    cosine_similarity(train_filled),
    index=train_matrix.index, columns=train_matrix.index
)
item_sim_df = pd.DataFrame(
    cosine_similarity(train_filled.T),
    index=train_matrix.columns, columns=train_matrix.columns
)

print(f"Train: {len(train_ratings):,} ratings | Test: {len(test_ratings):,} ratings")
print(f"Global mean rating: {global_mean:.3f}")


def predict_ub(user_id, item_id, top_k=10):
    if user_id not in user_sim_df.index or item_id not in train_matrix.columns:
        return global_mean
    neighbors = user_sim_df[user_id].drop(user_id, errors='ignore').nlargest(top_k)
    nr = train_matrix.loc[neighbors.index, item_id].dropna()
    if nr.empty:
        return global_mean
    sims = neighbors[nr.index]
    denom = sims.abs().sum()
    return (sims * nr).sum() / denom if denom > 0 else global_mean


def predict_ib(user_id, item_id):
    if item_id not in item_sim_df.index or user_id not in train_matrix.index:
        return global_mean
    rated = train_matrix.loc[user_id].dropna()
    common = rated.index.intersection(item_sim_df.columns)
    if common.empty:
        return global_mean
    sims = item_sim_df.loc[item_id, common]
    denom = sims.abs().sum()
    return (sims * rated[common]).sum() / denom if denom > 0 else global_mean


def recommend_for_eval(user_id, method='user', top_n=5):
    """Trả về list food_id gợi ý cho user — vectorized, không dùng Python loop."""
    if user_id not in train_matrix.index:
        return []
    rated      = train_matrix.loc[user_id].dropna().index
    candidates = train_matrix.columns.difference(rated)
    if candidates.empty:
        return []

    if method == 'user':
        neighbors = user_sim_df[user_id].drop(user_id, errors='ignore').nlargest(10)
        # Matrix ops thay cho Python loop: shape (K, n_candidates)
        nb_mat = train_matrix.loc[neighbors.index, candidates]
        sims   = neighbors.values[:, None]                    # (K, 1)
        mask   = nb_mat.notna().values
        num    = (nb_mat.fillna(0).values * sims).sum(axis=0)
        den    = (mask * np.abs(sims)).sum(axis=0)
        scores = pd.Series(np.where(den > 0, num / den, 0), index=candidates)

    else:
        liked  = train_matrix.loc[user_id].dropna()
        liked  = liked[liked >= 3.5]
        if liked.empty:
            return []
        cands  = candidates.intersection(item_sim_df.index)
        l_cols = liked.index.intersection(item_sim_df.columns)
        if cands.empty or l_cols.empty:
            return []
        # Matrix multiply: (n_cand, n_liked) @ (n_liked,) — không cần for loop
        S   = item_sim_df.loc[cands, l_cols].values
        lv  = liked[l_cols].values
        den = np.abs(S).sum(axis=1)
        scores = pd.Series(np.where(den > 0, S @ lv / den, 0), index=cands)

    return scores.nlargest(top_n).index.tolist()
```

```python
## Evaluation — RMSE, MAE, Precision@K, Recall@K

import time

# ── 1. RMSE / MAE ────────────────────────────────────────────────────────────
test_sample = test_ratings.sample(min(1000, len(test_ratings)), random_state=42)

# Dùng apply thay iterrows — tránh overhead Python loop từng dòng
t0 = time.time()
ub_preds = test_sample.apply(lambda r: predict_ub(r.userId, r.foodId), axis=1).values
ib_preds = test_sample.apply(lambda r: predict_ib(r.userId, r.foodId), axis=1).values
actuals  = test_sample['rating'].values
baseline = np.full(len(actuals), global_mean)
print(f"Prediction time: {time.time() - t0:.1f}s")

results = {
    'User-based':      (np.sqrt(mean_squared_error(actuals, ub_preds)), mean_absolute_error(actuals, ub_preds)),
    'Item-based':      (np.sqrt(mean_squared_error(actuals, ib_preds)), mean_absolute_error(actuals, ib_preds)),
    'Baseline (mean)': (np.sqrt(mean_squared_error(actuals, baseline)), mean_absolute_error(actuals, baseline)),
}

print(f"\n{'Method':<18} {'RMSE':>8} {'MAE':>8}")
print("─" * 36)
for method, (rmse, mae) in results.items():
    print(f"{method:<18} {rmse:>8.4f} {mae:>8.4f}")

# ── 2. Precision@K / Recall@K ─────────────────────────────────────────────────
def precision_recall_at_k(method='user', top_n=5, threshold=3.5, sample_users=60):
    np.random.seed(42)
    test_users = test_ratings['userId'].unique()
    eval_users = np.random.choice(test_users, min(sample_users, len(test_users)), replace=False)

    precisions, recalls = [], []
    for uid in eval_users:
        user_test = test_ratings[test_ratings['userId'] == uid]
        relevant  = set(user_test[user_test['rating'] >= threshold]['foodId'])
        if not relevant:
            continue
        recommended = set(recommend_for_eval(uid, method=method, top_n=top_n))
        if not recommended:
            continue
        hits = relevant & recommended
        precisions.append(len(hits) / top_n)
        recalls.append(len(hits) / len(relevant))

    p = np.mean(precisions) if precisions else 0.0
    r = np.mean(recalls)    if recalls    else 0.0
    return p, r

t0 = time.time()
ub_p, ub_r = precision_recall_at_k(method='user', top_n=5)
ib_p, ib_r = precision_recall_at_k(method='item', top_n=5)
print(f"\nPrecision/Recall time: {time.time() - t0:.1f}s")

print(f"\n{'Method':<18} {'Precision@5':>12} {'Recall@5':>10}")
print("─" * 42)
print(f"{'User-based':<18} {ub_p:>12.4f} {ub_r:>10.4f}")
print(f"{'Item-based':<18} {ib_p:>12.4f} {ib_r:>10.4f}")
```

## Export model artifacts cho FastAPI

```python
import joblib, os
from sklearn.metrics.pairwise import cosine_similarity as cos_sim

OUTPUT_DIR = 'model'
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Tính lại similarity từ FULL user_item (không dùng train_matrix — cái đó chỉ dùng để eval)
print("Computing similarity matrices from full data...")
ui_filled        = user_item.fillna(0)
user_sim_full    = pd.DataFrame(cos_sim(ui_filled),   index=user_item.index,   columns=user_item.index)
item_sim_full    = pd.DataFrame(cos_sim(ui_filled.T), index=user_item.columns, columns=user_item.columns)
global_mean_full = ratings['rating'].mean()

artifacts = {
    'user_sim':    user_sim_full,
    'item_sim':    item_sim_full,
    'user_item':   user_item,
    'foods':       foods[['food_id', 'dish_name', 'dish_type', 'calories',
                           'fat', 'fiber', 'sugar', 'protein',
                           'dish_tags', 'image_link']],
    'global_mean': global_mean_full,
}

path = os.path.join(OUTPUT_DIR, 'cf_artifacts.pkl')
joblib.dump(artifacts, path, compress=3)

size_mb = os.path.getsize(path) / 1024 / 1024
print(f"Saved  → {path}  ({size_mb:.1f} MB)")
print()
for k, v in artifacts.items():
    shape = v.shape if hasattr(v, 'shape') else v
    print(f"  {k:<12}: {shape}")
```

```python
## Tạo main.py cho FastAPI

fastapi_code = '''from fastapi import FastAPI, HTTPException
from contextlib import asynccontextmanager
import joblib, numpy as np, pandas as pd


@asynccontextmanager
async def lifespan(app: FastAPI):
    data = joblib.load("model/cf_artifacts.pkl")
    app.state.user_sim    = data["user_sim"]
    app.state.item_sim    = data["item_sim"]
    app.state.user_item   = data["user_item"]
    app.state.foods       = data["foods"].set_index("food_id")
    app.state.global_mean = data["global_mean"]
    yield

app = FastAPI(title="Vietnamese Food Recommender", lifespan=lifespan)


def _safe_div(num, den):
    out = np.zeros_like(num, dtype=float)
    mask = den > 0
    out[mask] = num[mask] / den[mask]
    return out


@app.get("/recommend/user/{user_id}")
def recommend_user_based(user_id: int, top_n: int = 5, top_k: int = 10):
    """Gợi ý món dựa trên user tương tự."""
    user_sim  = app.state.user_sim
    user_item = app.state.user_item
    foods     = app.state.foods

    if user_id not in user_sim.index:
        raise HTTPException(404, f"user_id {user_id} not found")

    rated      = user_item.loc[user_id].dropna().index
    candidates = user_item.columns.difference(rated)
    neighbors  = user_sim[user_id].drop(user_id, errors="ignore").nlargest(top_k)

    nb_mat = user_item.loc[neighbors.index, candidates]
    sims   = neighbors.values[:, None]
    mask   = nb_mat.notna().values
    num    = (nb_mat.fillna(0).values * sims).sum(axis=0)
    den    = (mask * np.abs(sims)).sum(axis=0)
    scores = pd.Series(_safe_div(num, den), index=candidates)

    top_ids = scores.nlargest(top_n).index
    return {
        "user_id": user_id,
        "method": "user-based",
        "recommendations": [
            {
                "food_id": int(fid),
                "dish_name": foods.loc[fid, "dish_name"],
                "score": round(float(scores[fid]), 4),
            }
            for fid in top_ids if fid in foods.index
        ],
    }


@app.get("/recommend/item/{user_id}")
def recommend_item_based(user_id: int, top_n: int = 5, threshold: float = 3.5):
    """Gợi ý món dựa trên món tương tự với món đã thích."""
    item_sim  = app.state.item_sim
    user_item = app.state.user_item
    foods     = app.state.foods

    if user_id not in user_item.index:
        raise HTTPException(404, f"user_id {user_id} not found")

    liked = user_item.loc[user_id].dropna()
    liked = liked[liked >= threshold]
    if liked.empty:
        raise HTTPException(400, "User has no highly-rated items to base recommendations on")

    rated      = user_item.loc[user_id].dropna().index
    candidates = user_item.columns.difference(rated)
    cands      = candidates.intersection(item_sim.index)
    l_cols     = liked.index.intersection(item_sim.columns)

    S   = item_sim.loc[cands, l_cols].values
    lv  = liked[l_cols].values
    den = np.abs(S).sum(axis=1)
    scores = pd.Series(_safe_div(S @ lv, den), index=cands)

    top_ids = scores.nlargest(top_n).index
    return {
        "user_id": user_id,
        "method": "item-based",
        "recommendations": [
            {
                "food_id": int(fid),
                "dish_name": foods.loc[fid, "dish_name"],
                "score": round(float(scores[fid]), 4),
            }
            for fid in top_ids if fid in foods.index
        ],
    }


@app.get("/foods/{food_id}")
def get_food(food_id: int):
    """Lấy thông tin chi tiết một món."""
    foods = app.state.foods
    if food_id not in foods.index:
        raise HTTPException(404, f"food_id {food_id} not found")
    return {"food_id": food_id, **foods.loc[food_id].to_dict()}
'''

with open("main.py", "w", encoding="utf-8") as f:
    f.write(fastapi_code.strip())

print("Wrote main.py")
print()
print("Cài dependencies:")
print("  pip install fastapi uvicorn joblib")
print()
print("Chạy server:")
print("  uvicorn main:app --reload")
print()
print("Endpoints:")
print("  GET /recommend/user/{user_id}?top_n=5&top_k=10")
print("  GET /recommend/item/{user_id}?top_n=5&threshold=3.5")
print("  GET /foods/{food_id}")
print("  GET /docs   ← Swagger UI tự động")
```
